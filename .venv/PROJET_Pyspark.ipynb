```python
from pyspark.sql import SparkSession

# Création ou récupération d'une session Spark
spark = SparkSession.builder \
    .appName("Analyse des crimes à Chicago") \
    .getOrCreate()

# Lecture du fichier CSV
df = spark.read.csv("Crimes_-_2001_to_Present_20251013.csv", header=True, inferSchema=True)

# Afficher les 20 premières lignes
df.show()
```

```python
# Le nombre de ligne
df.count()
```

```python
# Le type de mon données
type(df)
```

```python
# Le type des variables
df.printSchema()
```

Manipulation des données

```python
from pyspark.sql.functions import to_timestamp, col
```

```python
# Convertir la colonne Date en format Date
df = df.withColumn(
   "Date",
   to_timestamp(col("Date"), "MM/dd/yyyy hh:mm:ss a")
)

df.printSchema()
```

```python
# Afficher les 5 premières lignes
df.show(5)
```

```python
# Afficher les colonnes
df.columns
```

```python
# Selectionner la colonne Descriptions et afficher les 5 premières lignes
df.select(col("Description")).show(5)
```

```python
# Selectionner la colonne Date, Description, Arrest et afficher leurs 5 première ligne
df.select("Date", "Description", "Arrest").show(5)
```

```python
from pyspark.sql.functions import to_timestamp, lit
```

```python
# Créer une nouvelle colonne contenant que 1

df.withColumn(
   "One",
   lit(1)
).show(5)
```

```python
# Je veux les crimes signalés du 25-12-2023

one_day = df.filter(col("Date") == lit("2023-12-25"))
one_day.show()
```

```python
# Compter le nombre des crimes signalés le 25-12-2023
one_day.count()
```

```python
# Compter le nombre de ligne distinct dans mon DataFrame
df.distinct().count()
```

```python
# Grouper les résultats par Type de crime et afficher les 20 premières lignes
df.groupBy("Primary Type").count().show()
```

```python
# Le nombre de ligne distinct pour la colonne Arrest
df.select("Arrest").distinct().show()
```

```python
# Calcul du pourcentage du crime signalé de 2001 jusqu'a aujourd'hui
(df.filter(col("Arrest") == "true").count() / df.select("Arrest").count())*100
```

```python
# Les principaux lieux où les crimes sont signalés
n_location = df.groupBy("Location Description").count()
n_location.show()
```

```python
# Les trois principaux lieux où les crimes sont signalés
n_location.orderBy("count", ascending=False).show(3)
```

```python
# L'activité  criminelle la plus signalé
df.groupBy("Primary Type").count().orderBy("count", ascending=False).show()
```

```python
# L'activité  criminelle la moins signalé
df.groupBy("Primary Type").count().orderBy("count", ascending=True).show()
```

```python
df.select("Primary Type").show(30, truncate=False)
```

```python
# Le nombre de types de crime distinct
n_crime_type = df.select("Primary Type").distinct().count()
n_crime_type
```

```python
# Les activités non criminéls
df.select("Primary Type").distinct().show(n_crime_type, truncate=False)
```

```python
# Créer une DataFrame pour les activités non criminelles
nc = df.filter(
    (col("Primary Type") == "NON - CRIMINAL") | (col("Primary Type") == "NON-CRIMINAL") | (col("Primary Type") == "NON-CRIMINAL (SUBJECT SPECIFIED)"))
nc.show()
```

```python
# Compter le nombre de ligne
nc.count()
```

```python
# L'activité non criminelle la plus signalé
nc.groupBy("Description").count().orderBy("count", ascending=False).show()
```
